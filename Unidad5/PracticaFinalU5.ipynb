#### Carga y exploracion de datos

```python
import pandas as pd
df1 = pd.read_csv("grupo1.csv", encoding = "ISO-8859-1")
df1.head()

```

```python
df1.describe()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.hist(df1['Calificacion'], bins='sturges', color='skyblue', edgecolor='black')
plt.title("Histograma de calificacion")
plt.xlabel("Calificacion")
plt.ylabel("Frecuencia")
plt.show()
```

##### Modelo de Regresion Lineal **DF1**

-------------------------------

```python
from sklearn.linear_model import LinearRegression

x1 = df1[['HorasEstudio']]
y1 = df1['Calificacion']

modelo1 = LinearRegression()
modelo1.fit(x1, y1)

print("Pendiente: ", modelo1.coef_[0])
print("Intercepcion: ", modelo1.intercept_)
```

#####Coeficiente Pearson **DF1**

```python
from scipy.stats import pearsonr

x1 = df1['HorasEstudio']
y1 = df1['Calificacion']

coeficiente_correlacion, p_valor1 = pearsonr(x1, y1)

print("Coeficiente de correlacion de Pearson: ", coeficiente_correlacion)
print("Valor p: ", p_valor1)
```

##### Visualizacion de Resultados **DF1**

-------------------------------

```python
plt.scatter(x1, y1, color='blue', label='Datos')
plt.plot(x1, modelo1.predict(x1), color='red', linewidth=2, label='Regresion')
plt.xlabel('Horas Estudio')
plt.ylabel('Calificacion')
plt.title('Regresion Lineal Simple: Horas vs Calificacion')
plt.legend()
plt.show()
```

### Interpretacion de Resultados

1.-Explicar cuál es la interpretación del coeficiente de regresión en términos de la
relación entre el tiempo de estudio y las calificaciones finales

Bueno basicamente nos dice que por cada hora extra que un estudiante le dedique al estudio su calificacion final va a disminuir al rededor de un .1061 puntos, muy cercano a cero, ahora, esto nos dice que no tiene un efecto real sobre las calificaciones finales el tiempo de estudio en estos datos.

2.-¿Cómo se interpreta el coeficiente de correlación en términos de la fuerza y dirección de la relación entre las variables?

El signo negativo nos dice una relacion inversa, al estar tan porixmo a cero no es como tal una tendencia negativa real, por otro lado entre las dos variables no hay una relacion o casi inexistente, esto lo notamos en el diagrama donde los puntos se dispersan sin un seguimiento

3.-¿El modelo de regresión lineal es estadísticamente significativo?

El modelo estadisticamente no es significativo, la pequeña inclinacion de la linea de regresion no es de una relacion real entre las variables

4.- ¿Qué tan confiable es el modelo de regresión para predecir las calificaciones finales de los estudiantes basado en el tiempo de estudio?

Pracitcamente no es confiable, al sacar el coeficiente de determinacion nos .003 eso nos dice menos del 1% de la variabilidad total de calificaciones, al menos este modelo no aporta ventaja predictiva
