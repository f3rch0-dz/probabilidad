```python
import matplotlib.pyplot as plt
```

```python
x = [0,1,2,3,4,5]

p = [.10,.25,.30,.20,.10,.05]

print("Suma de probabilidades: ", sum(p))
print("\n")

plt.stem(x, p)
plt.title("Distribucion de Probabilidad")
plt.xlabel("Numero de celulas anormales")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()
```

```python
#Valor esperado

valor_esperado = 0

for i in range (len(x)):
  xi = x[i]
  pi = p[i]
  producto = xi * pi
  valor_esperado += producto
  print(f"x = {xi}, p = {pi}, x*p = {producto:.2f}")

print(f"\nValor esperado (esperanza): {valor_esperado:.2f}")
```

```python
# Varianza y desviacion estandar

import math

esperanza = 0
for i in range(len(x)):
  esperanza += x[i] * p[i]

esperanza_cuadrado = 0
for i in range(len(x)):
  esperanza_cuadrado += (x[i] ** 2) * p[i]

varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

print(f"E(X) = {esperanza:.2f}")
print(f"E(X'2) = {esperanza_cuadrado:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviacion estandar = {desviacion:.2f}")
```

```python
import math
import numpy as np
import matplotlib.pyplot as plt

F = np.cumsum(p)

for xi, fi in zip(x, F):
  print(f"F({xi}) = {fi:.2f}")

plt.step(x, F, where='post')
plt.title("Funcion de distribucion Acumulada (FDA)")
plt.xlabel("Numero de celulas anormales")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()
```

¿Cuál es el promedio esperado de células anormales por muestra?
El valor esperado es E(X) = 2.10 células anormales por muestra. Esto significa que a largo plazo, al revisar muchas muestras, te vas a encontrar en promedio con unas 2.1 células anormales por cada una.

¿Existe mucha variabilidad entre las muestras?
Sí, hay una variabilidad moderada. La desviación estándar es de 1.30 células (y la varianza de 1.69), lo que indica que los conteos reales en las muestras individuales se alejan o se dispersan más o menos 1.3 unidades respecto al promedio general.

¿Qué indica la función acumulada sobre la probabilidad de encontrar un número bajo o alto de células anormales?
La función acumulada muestra que lo más probable es encontrar números bajos o moderados. Por ejemplo, hay una probabilidad acumulada del 65 por ciento (F(2) = 0.65) de hallar 2 células anormales o menos. En cambio, la probabilidad de encontrar un número alto como 4 o 5 es bien baja, acumulando apenas el 5 por ciento para el caso máximo de 5 células.
