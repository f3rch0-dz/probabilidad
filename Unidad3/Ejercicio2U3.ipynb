```python
import matplotlib.pyplot as plt

x = [0, 10, 20, 50]
p = [0.50, 0.30, 0.15, 0.05]

print("Suma de probabilidades:", sum(p))
print("")

print(" Distribucion de probabilidad y grafica")
plt.stem(x, p)
plt.title("Distribucion de Probabilidad")
plt.xlabel("Monto del premio (pesos)")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()
```

```python
print(" Calculo de valor esperado, varianza y desviacion estandar")
import math

valor_esperado = 0
for i in range(len(x)):
    xi = x[i]
    pi = p[i]
    producto = xi * pi
    valor_esperado += producto
    print(f"x = {xi}, p = {pi}, x*p = {producto:.2f}")

esperanza = 0
for i in range(len(x)):
    esperanza += x[i] * p[i]

esperanza_cuadrado = 0
for i in range(len(x)):
    esperanza_cuadrado += (x[i] ** 2) * p[i]

varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

print(f"E(X) = {esperanza:.2f}")
print(f"E(X^2) = {esperanza_cuadrado:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviacion estandar = {desviacion:.2f}")
```

```python
print("Funcion de distribucion acumulada (FDA) y su grafica")
import numpy as np

F = np.cumsum(p)

for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")

plt.step(x, F, where='post')
plt.title("Funcion de distribucion Acumulada (FDA)")
plt.xlabel("Monto del premio (pesos)")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()
```

Qué significa el valor esperado en este contexto?
El valor esperado es de 8.50 pesos. Significa que, si juegas en la tómbola muchas veces a largo plazo, la ganancia promedio por cada boleto es de 8.50 pesos.

¿El juego es favorable para el jugador?
No es favorable si el boleto cuesta más de 8.50 pesos, ya que en promedio terminarías perdiendo dinero. Además, el 50 por ciento de las veces no te llevas nada (0 pesos) y el 80 por ciento de las veces ganas a lo sumo 10 pesos, por lo que la probabilidad de salir perdiendo o ganar muy poco es alta.
