```python
from statistics import *
```

```python
#Lista de calificaciones

grupos = [70, 80, 80, 90, 70, 100, 100, 94, 85, 79, 92]
print(grupos)
type(grupos)
```

```python
#Sacar la media de las calificaciones con una formula

promedio = sum(grupos) / len(grupos)
print(promedio)
```

```python
#Utilizar la funcion mean que esta en la libreria statistics para sacar el promedio

promedio2 = mean(grupos)
print(promedio2)
```

```python
#Determina la mediana

mediana = median(grupos)
print(mediana)
```

```python
#Determinar el valor de la moda con la funcion mode

moda = mode(grupos)
print(grupos)
print(moda)
```

```python
#En caso de que exista mas de una moda se puede remplazar funcion multimoda
#para sacar las multiples modas que puedan existir
moda2 = multimode(grupos)
print(moda2)
```

# Nuevo ejemplo

```python

#Definimos una variable llamada tiempos100mts de tipo lista que contiene los
#tiempos que tardaron los 18 alumnos para correr los 100mts
tiempos100mts = [15,10,17,20 ,14.69, 13.27, 18.71, 19.15, 20.65, 15.10, 17.20]

#Desplegamos los valores de la lista
print(tiempos100mts)
```

```python
#Usamos la funcion mean para determinar el tiempo promedio de la lista de valores

print("El tiempo promedio de los corredores es --->", mean(tiempos100mts));

```

```python
#Usamos la funcion median para determinar el tiempo promedio de la lista de valores

#definimos una variable llamada mediana por calcular la mediana de la lista con la funcion [median]
mediana = median(tiempos100mts);
print("La media de los tiempos de los corredores es --->", mediana)
```

```python
#Funcion mode para calcular la moda, definimos una variable llamada moda para calcular la moda de la lista con la funcion mode

moda = mode(tiempos100mts);
print("La moda de los tiempos es --->", moda);
```

```python
#multimode

multimode = multimode(tiempos100mts);
print("La moda de los tiempos es --->", multimode);

```

# Ejemplo 3
## Histogramas de 2 grupos


```python
# Importar para generar numeros aleatorios
import random
```

```python
fabrica1 = []
for i in range(100):
  numero_aleatorio = round( random.uniform(50, 100), 2)
  fabrica1.append(numero_aleatorio)
print(fabrica1)
```

```python
fabrica2 = []
for i in range(100):
  numero_aleatorio2 = round( random.uniform(70, 100), 2)
  fabrica2.append(numero_aleatorio2)
print(fabrica2)
```

```python
#Libreria de python para graficar valores
import matplotlib.pyplot as plt
```

```python
plt.hist(fabrica1)
plt.title("Producto diaria de la fabrica 1")

```

```python
plt.hist(fabrica2)
plt.title("Producto diaria de la fabrica 2")

```

```python
media1 = mean(fabrica1)
mediana1 = median(fabrica1)
moda1 = mode(fabrica1)

print("Valores de la fabrica 1")
print("La media es: ", media1)
print("La mediana es: ", mediana1)
print("La moda es: ", moda1)
```

```python
media2 = mean(fabrica1)
mediana2 = median(fabrica1)
moda2 = mode(fabrica1)

print("Valores de la fabrica 2")
print("La media es: ", media2)
print("La mediana es: ", mediana2)
print("La moda es: ", moda2)
```

```python
# generar 2 subplot de 1 renglon por 2 columnas para desplegar los 2 histogramas, una al lado de la otra
fig, (ax1, ax2) = plt.subplots(nrows =1, ncols=2)
```

```python
# generar 2 subplot de 1 renglon por 2 columnas para desplegar los 2 histogramas, una al lado de la otra
fig, (ax1, ax2) = plt.subplots(nrows =1, ncols=2, figsize=(10,5))
ax1.hist(fabrica1)
ax2.hist(fabrica2)
plt.show()
```

```python
#Valor minimo, maximo, rango

media1 = mean(fabrica1)
mediana1 = median(fabrica1)
moda1 = mode(fabrica1)
```

```python
plt.hist(fabrica1,bins=10, alpha=0.2, color='blue', edgecolor = 'black')
plt.axvline(media1, color='b', linestyle='dashed', linewidth=2, label='Media')
plt.axvline(mediana1, color='r', linestyle='dashed', linewidth=2, label='Mediana')
plt.axvline(moda1, color='y', linestyle='dashed', linewidth=2, label='Moda')

plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title("Produccion diaria de la fabrica 1")
plt.legend()
plt.show()
```

```python
media2 = mean(fabrica2)
mediana2 = median(fabrica2)
moda2 = mode(fabrica2)

plt.hist(fabrica2,bins=10, alpha=0.2, color='blue', edgecolor = 'black')
plt.axvline(media2, color='b', linestyle='dashed', linewidth=2, label='Media')
plt.axvline(mediana2, color='r', linestyle='dashed', linewidth=2, label='Mediana')
plt.axvline(moda2, color='y', linestyle='dashed', linewidth=2, label='Moda')

plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title("Produccion diaria de la fabrica 2")
plt.legend()
plt.show()
```
