# **Realizar un histograma de precios de los automoviles**

```python
# determinar mediante la regla de sturges la cantidad optima de bins o separadores

plt.hist(autos['price'], bins='sturges', color='skyblue', edgecolor='black')
plt.title("Histograma de Precios")
plt.xlabel("Precios")
plt.ylabel("Frecuencia")
plt.show()
```

## **Conclusion Histograma de Precios**

##### En el histograma determino que hay una diferencia muy grande de los carros que comienzan en 5,000 hasta aproximadamente los 9,000 USD, los cuales muestran una Frecuencia muy alta, despues la misma se mantiene en una forma en la que baja en un bloque muy similar desde los 9,000 hasta los 18,000 USD, despues de esto decae, pero ocurre un imprevisto, curiosamente despues de los 27,500 hasta los 38,000 hay una subida, para ya casi terminar esta baja nuevamente y un porcentaje subio desde los 41,000 hasta los 45,000, entonces concluyo que hay un mercado alto en los carros que rondan desde los 100,000 MX - 180,000 MX, no es que en los siguientes no haya mercado, pero hay un indice muy grande en este margen de precios.

# Realizar un diagrama de caja y bigotes que muestre la relacion estadistica entre la columna del precio del automovil **price** y tipo de vehiculo **body style**

```python
sns.set_style("whitegrid")

sns.boxplot(x="price", y="body-style", data=autos, palette='pastel')

plt.title("Relacion", fontsize=14, fontweight="bold")
plt.xlabel("Precios en los carros", fontsize=12)
plt.ylabel("Estilo de carro", fontsize=12)
```

## **Conclusion Histograma de Precios**

##### En el diagrama determino que los estilos de carrocería hardtop y convertible son los que presentan los precios más elevados del mercado, situándose sus medianas por encima de los 18,000 USD y alcanzando valores máximos de hasta 45,000 USD. Por el contrario, el estilo hatchback se posiciona como el segmento más accesible, concentrando la gran mayoría de sus precios por debajo de los 15,000 USD y teniendo la mediana más baja de la muestra. Por otro lado, la categoría sedan muestra una variabilidad y dispersión muy amplia en sus costos, entonces concluyo que los vehículos compactos o hatchback dominan el rango económico, mientras que los estilos hardtop y convertible pertenecen a un nicho de mayor poder adquisitivo, dejando al estilo sedan como la opción más versátil al cubrir casi todos los rangos de precio disponibles.

```python
sns.set_style("whitegrid")

sns.boxplot(x="price", y="drive-wheels", data=autos, palette='pastel')

plt.title("Relacion", fontsize=14, fontweight="bold")
plt.xlabel("Precios en los carros", fontsize=12)
plt.ylabel("Estilo de llantas", fontsize=12)
```

# **Conclusion histograma a relacion de Estilo de llantas y los Precios en los carros**

##### En el diagrama determino que los vehículos con tracción trasera rwd son considerablemente los más caros, situando su mediana cerca de los 17,000 USD y alcanzando los valores más altos del mercado hasta 45,000 USD. Por el contrario, los de tracción delantera fwd y cuatro ruedas 4wd son los más económicos, concentrando sus precios por debajo de los 12,000 USD, aunque en fwd se observan varios valores atípicos que superan los 15,000 USD.

```python
sns.set_style("whitegrid")

sns.boxplot(x="price", y="engine-location", data=autos, palette='pastel')

plt.title("Relacion", fontsize=14, fontweight="bold")
plt.xlabel("Precios en los carros", fontsize=12)
plt.ylabel("Ubicacion del motor", fontsize=12)
```

# **Conclusion del diagrama para la ubicacion del motor a su precio**

##### En el diagrama determino que la gran mayoría de los vehículos tienen el motor al frente front y se concentran en un rango económico mediana cercana a los 10,000 USD, aunque presentan bastantes valores atípicos puntos que superan los 30,000 USD. En contraste, los pocos autos con motor trasero rear son exclusivamente de gama alta, con precios muy elevados y uniformes entre los 32,000 y 37,000 USD.
