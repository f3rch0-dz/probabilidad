Es Binomial por que basicamente tiene un numero exacto de ensayos solo dos resultados posibles y una constante

```python
from scipy.stats import binom

n = 1000
p = 0.001

prob1 = binom.pmf(0, n, p)
print(f"Probabilidad de que ninguno sea defectuoso: {prob1:.4f}")

# Probabilidad de que tres sean defectuosos
prob2 = binom.pmf(3, n, p)
print(f"Probabilidad de que tres sean defectuosos: {prob2:.4f}")

# Probabilidad de que cuatro sean defectuosos
prob3 = binom.pmf(4, n, p)
print(f"Probabilidad de que cuatro sean defectuosos: {prob3:.4f}")
```
