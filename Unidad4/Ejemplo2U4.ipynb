```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

plt.figure(figsize=(6, 4))

#parametros de la distribucion normal
media = 70
desv_std = 4
total_estudiantes = 500

#variable para calcuo de la probabilidad de que un estudiante que pese menos de 75kg
peso_limite = 75

#primero necesitamos calcular la prob de que un estudiante pese menos de 75 utiz la funcion de dist acumulativa CDF de la distribucion normal
prob_menos_75kg = norm.cdf(peso_limite,media,desv_std)
```

```python
#calculo del numero de estudiantes que pesan menos de 75kg
estudiantes_menos_75kg = total_estudiantes * prob_menos_75kg

print("Probabilidad de que un estudiante pese menos de 75 kg:", prob_menos_75kg)
print("Numero de estudiantes que pesan menos de 75kg:", estudiantes_menos_75kg)

#graficar la distribucion normal

x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=(x <= peso_limite), color='skyblue', alpha=0.5)
plt.title('Probabilidad de que un estudiante pese menos de 75kg. \n $\mu=70$, $\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize= (6, 4))
#parametros de la distribucion normal
media = 70
desv_std = 4
peso_limite = 79
total_estudiantes = 500

#calculamos la probabilidad de que un estudiante pese menos o igual a 79
prob_menos_79kg = norm.cdf(peso_limite, media, desv_std)

#la prob de que un estudiante pese mas de 79kg es complementario
prob_mas_79kg = 1 - prob_menos_79kg

#calculamos el numero aproximado de estudiantes que pesan mas de 79kg
estudiantes_mas_79kg  = total_estudiantes * prob_mas_79kg

print("Probabilidad de que un estudiante pese mas de 79kg:", prob_mas_79kg)
print("Numero aproximado de estudiantes que pesan mas de 79kg:", estudiantes_mas_79kg)
```

```python
#calcular la funcion de densidad de probabilidad de la dsitribucion normal
x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, total_estudiantes)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=(x >= peso_limite), color='Skyblue', alpha=0.5)
plt.title('Probabilidad de que un estudiante pese mas de 79kg, \n $\mu=70$, $\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```
