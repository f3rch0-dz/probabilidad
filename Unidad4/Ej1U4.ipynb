```python
import scipy.stats as stats

#datos
estudiantes = 500
media = 70
desviacion = 4
una = 68
dos = 72

probabilidad = stats.norm.cdf(dos, loc=media, scale=desviacion) - stats.norm.cdf(una, loc=media, scale=desviacion)

cantidad_estudiantes = estudiantes * probabilidad

print("resultados")
print(f"1.Prob de pesar entre {una} kg y {dos} kg: {probabilidad:.4f} ({probabilidad * 100:.2f}%) ")
print(f"2. Cantidad estimada de estudiantes: {round(cantidad_estudiantes)} estudiantes (Valor exacto: {cantidad_estudiantes:.2f})")
```
