**Es de tipo poisson por que nos esta diciendo que el numero de personas ingresadas tiene una distribucion de probabilidad de poisson**

```python
from scipy.stats import poisson




media = 5


a = 2
probabilidad_dos = poisson.pmf(a, media)
print(f"a) La probabilidad de que ingresen exactamente {a} personas en un día es: {probabilidad_dos:.4f}")



b = 10
probabilidad_excede_10 = 1 - poisson.cdf(b, media)
print(f"b) La probabilidad de que ingresen más de {b} personas en un día es: {probabilidad_excede_10:.4f}")
```

a) La probabilidad de que el número de personas ingresadas sea exactamente dos es de aproximadamente 0.0842, es decir, un 8.42%.

b) No es probable que el número exceda de 10. La probabilidad de que ingresen más de 10 personas en un solo día es de aproximadamente 0.0137 (1.37%), lo cual es un porcentaje muy bajo tomando en cuenta que el promedio diario es de solo 5 personas.
